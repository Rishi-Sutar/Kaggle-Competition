```python
def summarize(obs):
    return {
        "step": obs.get("step"),
        "day": obs.get("day"),
        "hour": obs.get("hour"),

        "p0_money": obs["farms"][0]["money"],
        "p1_money": obs["farms"][1]["money"],

        "p0_farmer": tuple(obs["farms"][0]["farmer"]),
        "p1_farmer": tuple(obs["farms"][1]["farmer"]),

        "p0_hands": len(obs["farms"][0]["hands"]),
        "p1_hands": len(obs["farms"][1]["hands"]),

        "shops": tuple(obs["town"]["unlocked_shops"]),

        "prices": dict(obs["market"]["prices"]),
        "market_inventory": dict(obs["market"]["inventory"]),
    }
```

```python
history = []

def experiment_agent(obs):
    history.append(summarize(obs))

    return {
        "farmer": ["PASS"],
        "market": [],
    }
```

```python
env = make(
    "kaggriculture",
    configuration={
        "episodeSteps": 48,
        "seed": 42,
    },
    debug=True,
)

result = env.run([
    experiment_agent,
    experiment_agent,
])
```
